```python
def map_team_names(df, team_column):
    """
    Maps team names to standardized values before encoding.

    Parameters:
    df (pd.DataFrame): The DataFrame containing the team column.
    team_column (str): The name of the column containing team names.

    Returns:
    pd.DataFrame: A DataFrame with standardized team names.
    """
    team_mapping = {
        "LAL": "Los Angeles Lakers",
        "LAC": "Los Angeles Clippers",
        "GSW": "Golden State Warriors",
        "BOS": "Boston Celtics",
        "MIA": "Miami Heat",
        "CHI": "Chicago Bulls",
        "NYK": "New York Knicks",
        "PHI": "Philadelphia 76ers",
        "MIL": "Milwaukee Bucks",
        "HOU": "Houston Rockets",
        "SAS": "San Antonio Spurs",
        "TOR": "Toronto Raptors",
        "DAL": "Dallas Mavericks",
        "DEN": "Denver Nuggets",
        "UTA": "Utah Jazz",
        "OKC": "Oklahoma City Thunder",
        "POR": "Portland Trail Blazers",
        "PHX": "Phoenix Suns",
        "NOP": "New Orleans Pelicans",
        "MEM": "Memphis Grizzlies",
        "BKN": "Brooklyn Nets",
        "SAC": "Sacramento Kings",
        "IND": "Indiana Pacers",
        "WAS": "Washington Wizards",
        "CLE": "Cleveland Cavaliers",
        "ATL": "Atlanta Hawks",
        "CHA": "Charlotte Hornets",
        "DET": "Detroit Pistons",
        "ORL": "Orlando Magic",
        "MIN": "Minnesota Timberwolves",
        "CHO": "Charlotte Hornets",
        "BRK": "Brooklyn Nets",
        "PHO": "Phoenix Suns",
    }

    # Apply the mapping to the team column
    df[team_column] = df[team_column].map(team_mapping).fillna(df[team_column])  # Keep original if not in mapping

    return df


```

# Clean 'player_stats' dataframe
### Columns Affected
- All columns: (lowercase)
- year : Use base year + 1
  - `ex. 2012 - 13, uses 2012 + 1 to use the prominent season year`
- season typ : removes `%20` from between values

```python
def clean_data(df):
    # Extracting the first part before the '-' and ensuring it's numeric
    df['year'] = df['year'].str.split('-').str[0]
    df['year'] = pd.to_numeric(df['year'], errors='coerce')

    # Drop rows with NaN in 'year' or fill with a default value (e.g., current year)
    df = df.dropna(subset=['year'])
    df['year'] = df['year'].astype(int)

    # Increment the year by 1 to match player_stats
    df['year'] = df['year'] + 1

    # Clean up Season_type column
    df['Season_type'] = df['Season_type'].str.replace('%20', " ")

    # Convert all column names to lowercase
    df.columns = df.columns.str.lower()

    # Drop ast_tov and stl/tov cols
    df = df.drop(columns=['ast_tov', 'stl_tov'])

    return df
```

```python
from sklearn.preprocessing import LabelEncoder

# Dictionary to store encoders for decoding later
encoders = {}

def encode_categorical_columns(df, columns):
    """
    Encodes specified categorical columns in a DataFrame.

    Parameters:
    - df (pd.DataFrame): DataFrame containing the categorical columns.
    - columns (list): List of column names to encode.

    Returns:
    - pd.DataFrame: DataFrame with encoded categorical columns.
    """
    global encoders  # Store encoders globally for decoding
    df_encoded = df.copy()

    for column in columns:
        if column in df_encoded.columns and len(df_encoded[column]) > 2:
            le = LabelEncoder()
            df_encoded[column] = le.fit_transform(df_encoded[column])
            encoders[column] = le  # Save the encoder for later decoding
        else:
            unique_values = df_encoded[column].unique()
            if len(unique_values) == 2:
              mapping = {unique_values[0]: 0, unique_values[1]: 1}
              encoders[column] = mapping
              df_encoded[column] = df_encoded[column].map(mapping)

    return df_encoded

def decode_categorical_columns(df, columns):
    """
    Decodes previously encoded categorical columns in a DataFrame.

    Parameters:
    - df (pd.DataFrame): DataFrame with encoded categorical columns.
    - columns (list): List of column names to decode.

    Returns:
    - pd.DataFrame: DataFrame with decoded categorical columns.
    """
    df_decoded = df.copy()

    for column in columns:
        if column in encoders and column in df_decoded.columns:
            le = encoders[column]
            df_decoded[column] = le.inverse_transform(df_decoded[column])

    return df_decoded
```

# Create dataframe for nba logos
### Create a dataframe of nba abbreviations and map them to each image in the dataset

```python
import pandas as pd
from PIL import Image
import numpy as np
import os
```

```python
# Change directory to the folder containing the images
image_folder = '/content/nba_logos/nba_logos'
os.chdir(image_folder)

# Get all image file names
image_files = [f for f in os.listdir() if f.endswith(('png', 'jpg', 'jpeg', 'gif'))]

# Create an empty list to hold image data
image_data = []

# Loop through each image file
for image_file in image_files:
    # Open image using PIL
    image = Image.open(image_file)

    # Get image properties
    file_name = image_file
    file_path = os.path.join(image_folder, image_file)
    width, height = image.size
    mode = image.mode

    # Optional: Convert image to numpy array
    image_array = np.array(image)

    # Append data to the list
    image_data.append({
        'File Name': file_name,
        'File Path': file_path,
        'Width': width,
        'Height': height,
        'Mode': mode,
        'Image Array': image_array
    })

# Create a DataFrame from the list
df_images = pd.DataFrame(image_data)

# Display the DataFrame
df_images.head()
```
